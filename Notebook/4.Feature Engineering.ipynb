```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Data/Raw/Telco-Customer-Churn.csv")
```

```python
df.head()
```

```python
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)
```

```python
print(df.columns)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(' ', '_')
)
```

```python
print(df.columns)
```

```python
df['Total_Charges'] = pd.to_numeric(
    df['Total_Charges'],
    errors='coerce'
)
```

```python
df.dropna(inplace=True)
```

```python
df['AvgRevenuePerMonth'] = (
    df['Total_Charges'] /
    (df['Tenure_Months'] + 1)
)
```

```python
df['LongTermCustomer'] = (
    df['Tenure_Months'] > 24
).astype(int)
```

```python
df['HighMonthlyCharges'] = (
    df['Monthly_Charges'] >
    df['Monthly_Charges'].median()
).astype(int)
```

```python
streaming_cols = [
    'Streaming_TV',
    'Streaming_Movies'
]

df['StreamingServicesCount'] = (
    (df[streaming_cols] == 'Yes')
    .sum(axis=1)
)
```

```python
df['Churn_Label'] = df['Churn_Label'].map({
    'Yes': 1,
    'No': 0
})
```

```python
df[[
    'AvgRevenuePerMonth',
    'LongTermCustomer',
    'HighMonthlyCharges',
    'StreamingServicesCount',
    'Churn_Label'
]].head()
```

```python
df.to_csv(
    "Churn_Feature_Engineered.csv",
    index=False
)
```
